```python
import pandas as pd
import sys
from transformers import DistilBertTokenizer
sys.path.append("../../splade_sans/learned-sparse-retrieval-1.0.0")
from lsr.transformer import LSR
import numpy as np
from pyterrier.model import add_ranks
import pyterrier_alpha as pta
```

```python
ohsumed_docs = pd.read_pickle("/nfs/primary/sas_reranker/ohsumed_docs_w_t5base_sensitivity.pkl")
queries = pd.read_pickle("/nfs/primary/graph_adaptive_reranking/queries.pkl")
qrels = pd.read_pickle("/nfs/primary/graph_adaptive_reranking/qrels.pkl")
```

```python
docs_in_qrels = pd.merge(ohsumed_docs, qrels, on = "docno")
docs_in_qrels = docs_in_qrels[docs_in_qrels["label"] == 2]
```

```python
sensitive_docs = docs_in_qrels[docs_in_qrels["sensitivity"] == 1].text.tolist()
non_sensitive_docs = docs_in_qrels[docs_in_qrels["sensitivity"] == 0].text.tolist()
```

```python
distil_model_path = "/nfs/primary/SPLADE/splade_class/lsr_package/outputs/hn_t5_retrain_scratch/model"
distil_model = LSR(distil_model_path)

relevance_model_path = "/nfs/primary/SPLADE/splade_class/lsr_package/outputs/splade_ohsumed_multiple_negative/model"
relevance_model = LSR(relevance_model_path)
```

```python
rel_sens = relevance_model.encode_docs(sensitive_docs, out_fmt = "np")
res_non = relevance_model.encode_docs(non_sensitive_docs, out_fmt = "np")
distil_sens = distil_model.encode_docs(sensitive_docs, out_fmt = "np")
distil_non = distil_model.encode_docs(non_sensitive_docs, out_fmt = "np")
```

```python
import torch
import torch.nn.functional as F

def rmse(tensor1, tensor2):
    return torch.sqrt(F.mse_loss(tensor1, tensor2))

rel_sens = torch.from_numpy(rel_sens)
distil_sens = torch.from_numpy(distil_sens)
rel_non = torch.from_numpy(res_non)
distil_non = torch.from_numpy(distil_non)

rmse_sens = rmse(rel_sens, distil_sens)
rmse_non = rmse(rel_non, distil_non)

print(f"RMSE for Sensitive Docs: {rmse_sens.item():.4f}")
print(f"RMSE for Non-Sensitive Docs: {rmse_non.item():.4f}")
```

```python
per_token_mean_sensitive = torch.mean(rel_sens - distil_sens)
per_token_mean_non = torch.mean(rel_non - distil_non)

per_token_mean_sensitive, per_token_mean_non
```

```python
def get_rbo(df1, df2):
    values = df1.mean(axis = 0)
    indexes = np.arange(len(df1.mean(axis = 0)))
    
    rel_sens_df = add_ranks(pd.DataFrame(
        {
            "qid" : [1] * len(indexes),
            "docno" : indexes,
            "score" : values
        }
    ))
    
    values = df2.mean(axis = 0)
    indexes = np.arange(len(df2.mean(axis = 0)))
    
    rel_non_df = add_ranks(pd.DataFrame(
        {
            "qid" : [1] * len(indexes),
            "docno" : indexes,
            "score" : values
        }
    ))
    
    rbo = pta.rbo(rel_sens_df, rel_non_df)
    print(rbo)
    return

get_rbo(rel_sens, distil_sens)
get_rbo(rel_non, distil_non)
```

```python
max_query_rep_relevance = relevance_model.encode_queries([queries[queries["qid"] == "53"]["query"].iloc[0]], out_fmt = "np")
max_query_rep_distil = distil_model.encode_queries([queries[queries["qid"] == "53"]["query"].iloc[0]], out_fmt = "np")

rel_sens_max_query = max_query_rep_relevance * np.array(rel_sens)
rel_non_max_query_rep = max_query_rep_relevance * np.array(rel_non)

distil_sens_max_query_rep_relevance = max_query_rep_distil * np.array(distil_sens)
distil_non_max_query_rep_distil = max_query_rep_distil * np.array(distil_non)
```

```python
min_query_rep_relevance = relevance_model.encode_queries([queries[queries["qid"] == "10"]["query"].iloc[0]], out_fmt = "np")
min_query_rep_distil = distil_model.encode_queries([queries[queries["qid"] == "10"]["query"].iloc[0]], out_fmt = "np")

rel_sens_min_query = min_query_rep_relevance * np.array(rel_sens)
rel_non_min_query_rep = min_query_rep_relevance * np.array(rel_non)

distil_sens_min_query_rep_relevance = min_query_rep_distil * np.array(distil_sens)
distil_non_min_query_rep_distil = min_query_rep_distil * np.array(distil_non)
```

```python
get_rbo(rel_sens_max_query, distil_sens_max_query_rep_relevance)
get_rbo(rel_non_max_query_rep, distil_non_max_query_rep_distil)
print()
get_rbo(rel_sens_min_query, distil_sens_min_query_rep_relevance)
get_rbo(rel_non_min_query_rep, distil_non_min_query_rep_distil)
```
