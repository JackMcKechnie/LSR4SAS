```python
import pandas as pd
```

```python
ohsumed_docs = pd.read_pickle("/nfs/primary/sas_reranker/ohsumed_docs_w_t5base_sensitivity.pkl")
queries = pd.read_pickle("/nfs/primary/graph_adaptive_reranking/queries.pkl")
qrels = pd.read_pickle("/nfs/primary/graph_adaptive_reranking/qrels.pkl")
```

```python
qrels = pd.merge(qrels, ohsumed_docs[["docno", "sensitivity"]], on = "docno") 
```

```python
# per-query counts
per_qid = qrels.groupby("qid").apply(
    lambda df: pd.Series({
        "relevant_sensitive": ((df["is_relevant"]) & (df["sensitivity"] == 1)).sum(),
        "relevant_non_sensitive": ((df["is_relevant"]) & (df["sensitivity"] == 0)).sum(),
    })
)

# averages
avg_rel_sens = round(per_qid["relevant_sensitive"].mean(), 4)
avg_rel_nonsens = round(per_qid["relevant_non_sensitive"].mean(), 4)

# standard deviations
std_rel_sens = round(per_qid["relevant_sensitive"].std(), 4)
std_rel_nonsens = round(per_qid["relevant_non_sensitive"].std(), 4)

avg_rel_sens, std_rel_sens, avg_rel_nonsens, std_rel_nonsens
```

```python
min_rel_sens = int(per_qid["relevant_sensitive"].min())
max_rel_sens = int(per_qid["relevant_sensitive"].max())

min_rel_sens, max_rel_sens
```
