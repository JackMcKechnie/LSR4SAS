```python
import pandas as pd
import json
```

```python
full_df = pd.read_csv("./training_data_w_qids.csv")
```

```python
qrels = {}
for _, row in full_df.iterrows():
    qid = str(row['qid'])
    doc_id = str(row['medline_ui'])
    if qid not in qrels:
        qrels[qid] = []
    qrels[qid].append(doc_id)
```

```python
qrels_path = "./training_qrels.json"
with open(qrels_path, 'w') as json_file:
    json.dump(qrels, json_file, indent=4)
```

```python
import json
strategy1 = pd.read_csv("sans_strat3_7neg.tsv", sep = "\t", header = None)
strategy1.columns = ["qid", "pos_docno", "neg_docno"]
qid_dict = (
    strategy1.groupby(["qid", "pos_docno"])["neg_docno"]
    .apply(lambda x: [x.name[1]] + x.tolist())  # Start with pos_docno, then up to 8 neg_docnos
    .reset_index(level=1, drop=True)
    .to_dict()
)

qid_dict = {int(k): [int(v) for v in vals] for k, vals in qid_dict.items()}


with open("strategy3_training_qrels.json", "w") as f:
    json.dump(qid_dict, f, indent=4)
```
