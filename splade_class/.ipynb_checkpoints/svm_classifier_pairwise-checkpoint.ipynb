```python
import pandas as pd
import numpy as np
```

```python
predictions = pd.read_csv("./predictions_with_probabilities.csv")
predictions.docno = predictions.docno.astype(str)
docs = pd.read_pickle("/nfs/primary/sas_reranker/ohsumed_docs_w_t5base_sensitivity.pkl")
```

```python
results = pd.merge(docs, predictions, on = "docno").sort_values("probability", ascending = False)
```

```python
# Extract probability values based on sensitivity
sensitive_probs = results.loc[results['sensitivity'] == 1, 'probability'].values
non_sensitive_probs = results.loc[results['sensitivity'] == 0, 'probability'].values

# Broadcasting comparison: Create a 2D array where each sensitive doc is compared to all non-sensitive docs
count = np.sum(sensitive_probs[:, None] > non_sensitive_probs)
total_pairs = len(sensitive_probs) * len(non_sensitive_probs)

print("Percentage of pair where sensitive document is ranked above a non-sensitive document", round(count / total_pairs, 4))
```
